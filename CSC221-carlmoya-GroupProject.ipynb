### This code will create a dataframe and a csv file both populated by fruit name and calorie data from the fruityvice api

```python
# Import requests to make get calls to the api
import requests

# Import pandas for data framing and csv exporting
import pandas as pd
```

### Store the fruityvice api url

```python
fruit_api = "https://www.fruityvice.com/api/fruit/all"
```

### Send a get request to the fruityvice api url

```python
api_response = requests.get(fruit_api)
```

### Convert the json response into a list of fruit dictionaries

```python
fruit_data = api_response.json()
```

### Create a dataframe containing only each fruit's name and calorie value

```python
fruit_df = pd.DataFrame([{"Name": fruit["name"], "Calories": fruit["nutritions"]["calories"]} for fruit in fruit_data])
```

### Save the fruit data frame to fruit_data.csv without row numbers

```python
fruit_df.to_csv("fruit_data.csv", index = False)
```
